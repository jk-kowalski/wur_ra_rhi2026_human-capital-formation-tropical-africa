```python
# packages
import geopandas as gpd
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np
from pathlib import Path
from scipy.stats import gaussian_kde
```

## Birthplace/primary - series per cohort

```python
# ---- paths ----
input_file = r"C:\Users\kubak\Documents\research_assistant_wur\research_assistant_rhi2026_africa\3_outputs\geom_edu_birthplace_8kenya.geojson"
output_dir = Path(r"C:\Users\kubak\Documents\research_assistant_wur\research_assistant_rhi2026_africa\3_outputs\kenya\birthplace_primary_maps")
output_dir.mkdir(exist_ok=True)

# ---- load data ----
gdf = gpd.read_file(input_file)

print(gdf.columns.tolist())

gdf["cohort"] = gdf["cohort"].astype("Int64")
gdf["primary_educ"] = gdf["primary_educ"].astype(float)
gdf = gdf[gdf["cohort"].notna() & gdf["primary_educ"].notna()].copy()

# Fixed scale across all maps
vmin = gdf["primary_educ"].min()
vmax = gdf["primary_educ"].max()

for cohort in sorted(gdf["cohort"].unique()):
    subset = gdf[gdf["cohort"] == cohort].copy()
    values = subset["primary_educ"].dropna().values

    fig, ax = plt.subplots(figsize=(8, 10))

    # ---- map ----
    subset.plot(
        column="primary_educ",
        cmap="viridis",
        linewidth=0.4,
        edgecolor="black",
        legend=False,
        vmin=vmin,
        vmax=vmax,
        ax=ax
    )

    ax.set_title(f"Kenya birthplace-based primary education, cohort {cohort}")
    ax.set_axis_off()

    # ---- horizontal colorbar ----
    norm = mpl.colors.Normalize(vmin=vmin, vmax=vmax)
    sm = mpl.cm.ScalarMappable(norm=norm, cmap="viridis")
    sm.set_array([])

    cbar = fig.colorbar(
        sm,
        ax=ax,
        orientation="horizontal",
        fraction=0.05,
        pad=0.04
    )
    cbar.set_label("Primary education share")

    # ---- wave axis above the colorbar ----
    # Get the colorbar position in figure coordinates
    cb_pos = cbar.ax.get_position()

    # Add a small axis directly above the colorbar
    wave_ax = fig.add_axes([
        cb_pos.x0,                # left
        cb_pos.y1 + 0.005,        # bottom
        cb_pos.width,             # width
        0.05                      # height
    ])

    mean_val = values.mean()
    min_val = values.min()
    max_val = values.max()

    fig.text(
        0.5, 0.02,
        f"Mean: {mean_val:.3f}   Min: {min_val:.3f}   Max: {max_val:.3f}",
        ha='center',
        fontsize=10
    )

    # Histogram-based smooth-ish wave
    kde = gaussian_kde(values)
    x = np.linspace(vmin, vmax, 300)
    y = kde(x)

    wave_ax.plot(x, y, linewidth=1.5)
    wave_ax.fill_between(x, y, alpha=0.25)

    # Match wave axis to colorbar scale
    wave_ax.set_xlim(vmin, vmax)
    wave_ax.set_xticks([])
    wave_ax.set_yticks([])

    # Clean appearance
    for spine in wave_ax.spines.values():
        spine.set_visible(False)

    out_file = output_dir / f"birthplace_primary_educ_{cohort}.png"
    plt.savefig(out_file, dpi=300, bbox_inches="tight")
    plt.close()

print("Done. Maps saved to:", output_dir)
```

```python
from pathlib import Path
import geopandas as gpd
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from scipy.stats import gaussian_kde
```

```python
from pathlib import Path
import geopandas as gpd
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from scipy.stats import gaussian_kde

# ---- paths ----
input_file = r"C:\Users\kubak\Documents\research_assistant_wur\research_assistant_rhi2026_africa\3_outputs\geom_edu_birthplace_8kenya_diff.geojson"
base_dir = Path(r"C:\Users\kubak\Documents\research_assistant_wur\research_assistant_rhi2026_africa\3_outputs\kenya\birthplace_male_maps")
base_dir.mkdir(parents=True, exist_ok=True)

# ---- load data ----
gdf = gpd.read_file(input_file)
print(gdf.columns.tolist())

# ---- clean fields ----
gdf["cohort"] = pd.to_numeric(gdf["cohort"], errors="coerce").astype("Int64")

male_columns = {
    "primary_male": "Primary education share",
    "higher_male": "Higher education share",
    "tertiary_male": "Tertiary education share"
}

for col in male_columns:
    gdf[col] = pd.to_numeric(gdf[col], errors="coerce")

gdf = gdf[gdf["cohort"].notna()].copy()

for edu_col, legend_label in male_columns.items():
    if edu_col not in gdf.columns:
        print(f"Skipping missing column: {edu_col}")
        continue

    # ---- create subfolder per education level ----
    edu_dir = base_dir / edu_col.replace("_male", "")
    edu_dir.mkdir(parents=True, exist_ok=True)

    edu_gdf = gdf[gdf[edu_col].notna()].copy()

    if edu_gdf.empty:
        print(f"No data for {edu_col}")
        continue

    # ---- education-specific scale ----
    values_all = edu_gdf[edu_col].dropna().values
    vmin = values_all.min()
    vmax = values_all.max()

    if vmin == vmax:
        vmax = vmin + 1e-6

    norm = mpl.colors.Normalize(vmin=vmin, vmax=vmax)
    print(f"{edu_col}: vmin={vmin:.6f}, vmax={vmax:.6f}")

    for cohort in sorted(edu_gdf["cohort"].dropna().unique()):
        subset = edu_gdf[edu_gdf["cohort"] == cohort].copy()

        if subset.empty:
            continue

        values = subset[edu_col].dropna().values
        if len(values) == 0:
            continue

        fig, ax = plt.subplots(figsize=(8, 10))

        # ---- map ----
        subset.plot(
            column=edu_col,
            cmap="viridis",
            linewidth=0.4,
            edgecolor="black",
            legend=False,
            norm=norm,
            ax=ax
        )

        ax.set_title(f"Kenya birthplace-based male {legend_label.lower()}, cohort {cohort}")
        ax.set_axis_off()

        # ---- horizontal colorbar ----
        sm = mpl.cm.ScalarMappable(norm=norm, cmap="viridis")
        sm.set_array([])

        cbar = fig.colorbar(
            sm,
            ax=ax,
            orientation="horizontal",
            fraction=0.05,
            pad=0.04
        )
        cbar.set_label(legend_label)

        # ---- wave axis above the colorbar ----
        cb_pos = cbar.ax.get_position()
        wave_ax = fig.add_axes([
            cb_pos.x0,
            cb_pos.y1 + 0.005,
            cb_pos.width,
            0.05
        ])

        mean_val = values.mean()
        min_val = values.min()
        max_val = values.max()

        fig.text(
            0.5, 0.02,
            f"Mean: {mean_val:.3f}   Min: {min_val:.3f}   Max: {max_val:.3f}",
            ha="center",
            fontsize=10
        )

        if len(values) > 1 and not np.all(values == values[0]):
            kde = gaussian_kde(values)
            x = np.linspace(vmin, vmax, 300)
            y = kde(x)

            wave_ax.plot(x, y, linewidth=1.5)
            wave_ax.fill_between(x, y, alpha=0.25)

        wave_ax.set_xlim(vmin, vmax)
        wave_ax.set_xticks([])
        wave_ax.set_yticks([])

        for spine in wave_ax.spines.values():
            spine.set_visible(False)

        out_file = edu_dir / f"{edu_col}_{cohort}.png"
        plt.savefig(out_file, dpi=300, bbox_inches="tight")
        plt.close()

print("Done. Maps saved to:", base_dir)
```

```python
from pathlib import Path
import geopandas as gpd
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from scipy.stats import gaussian_kde

# ---- paths ----
input_file = r"C:\Users\kubak\Documents\research_assistant_wur\research_assistant_rhi2026_africa\3_outputs\geom_edu_birthplace_8kenya_diff.geojson"
base_dir = Path(r"C:\Users\kubak\Documents\research_assistant_wur\research_assistant_rhi2026_africa\3_outputs\kenya\birthplace_gender_diff_maps")
base_dir.mkdir(parents=True, exist_ok=True)

# ---- load data ----
gdf = gpd.read_file(input_file)
print(gdf.columns.tolist())

# ---- clean fields ----
gdf["cohort"] = pd.to_numeric(gdf["cohort"], errors="coerce").astype("Int64")

diff_columns = {
    "primary_male_minus_female": "Primary education difference (male - female)",
    "higher_male_minus_female": "Higher education difference (male - female)",
    "tertiary_male_minus_female": "Tertiary education difference (male - female)"
}

for col in diff_columns:
    gdf[col] = pd.to_numeric(gdf[col], errors="coerce")

gdf = gdf[gdf["cohort"].notna()].copy()

for edu_col, legend_label in diff_columns.items():
    if edu_col not in gdf.columns:
        print(f"Skipping missing column: {edu_col}")
        continue

    edu_dir = base_dir / edu_col.replace("_male_minus_female", "")
    edu_dir.mkdir(parents=True, exist_ok=True)

    edu_gdf = gdf[gdf[edu_col].notna()].copy()

    if edu_gdf.empty:
        print(f"No data for {edu_col}")
        continue

    # ---- education-specific symmetric scale ----
    values_all = edu_gdf[edu_col].dropna().values
    abs_max = np.nanmax(np.abs(values_all))

    # fallback in case all values are exactly zero
    if abs_max == 0:
        abs_max = 1e-6

    norm = mpl.colors.TwoSlopeNorm(vmin=-abs_max, vcenter=0, vmax=abs_max)

    print(f"{edu_col}: vmin={-abs_max:.6f}, vmax={abs_max:.6f}")

    for cohort in sorted(edu_gdf["cohort"].dropna().unique()):
        subset = edu_gdf[edu_gdf["cohort"] == cohort].copy()

        if subset.empty:
            continue

        values = subset[edu_col].dropna().values
        if len(values) == 0:
            continue

        fig, ax = plt.subplots(figsize=(8, 10))

        # ---- map ----
        subset.plot(
            column=edu_col,
            cmap="RdBu_r",
            linewidth=0.4,
            edgecolor="black",
            legend=False,
            norm=norm,
            ax=ax
        )

        ax.set_title(f"Kenya birthplace-based gender difference, {legend_label.lower()}, cohort {cohort}")
        ax.set_axis_off()

        # ---- horizontal colorbar ----
        sm = mpl.cm.ScalarMappable(norm=norm, cmap="RdBu_r")
        sm.set_array([])

        cbar = fig.colorbar(
            sm,
            ax=ax,
            orientation="horizontal",
            fraction=0.05,
            pad=0.04
        )
        cbar.set_label(legend_label)

        # ---- wave axis above the colorbar ----
        cb_pos = cbar.ax.get_position()
        wave_ax = fig.add_axes([
            cb_pos.x0,
            cb_pos.y1 + 0.005,
            cb_pos.width,
            0.05
        ])

        mean_val = values.mean()
        min_val = values.min()
        max_val = values.max()

        fig.text(
            0.5, 0.02,
            f"Mean: {mean_val:.6f}   Min: {min_val:.6f}   Max: {max_val:.6f}",
            ha="center",
            fontsize=10
        )

        if len(values) > 1 and not np.all(values == values[0]):
            kde = gaussian_kde(values)
            x = np.linspace(-abs_max, abs_max, 300)
            y = kde(x)

            wave_ax.plot(x, y, linewidth=1.5)
            wave_ax.fill_between(x, y, alpha=0.25)

        wave_ax.set_xlim(-abs_max, abs_max)
        wave_ax.set_xticks([])
        wave_ax.set_yticks([])

        for spine in wave_ax.spines.values():
            spine.set_visible(False)

        out_file = edu_dir / f"{edu_col}_{cohort}.png"
        plt.savefig(out_file, dpi=300, bbox_inches="tight")
        plt.close()

print("Done. Maps saved to:", base_dir)
```
