```python
# packages
import geopandas as gpd
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np
from pathlib import Path
from scipy.stats import gaussian_kde
```

## Birthplace/primary - series per cohort

```python
# ---- paths ----
input_file = r"C:\Users\kubak\Documents\research_assistant_wur\research_assistant_rhi2026_africa\3_outputs\10mali\geom_edu_birthplace_10mali.geojson"
output_dir = Path(r"C:\Users\kubak\Documents\research_assistant_wur\research_assistant_rhi2026_africa\3_outputs\10mali\birthplace_primary_maps")
output_dir.mkdir(exist_ok=True)

# ---- load data ----
gdf = gpd.read_file(input_file)

print(gdf.columns.tolist())

gdf["cohort"] = gdf["cohort"].astype("Int64")
gdf["primary_educ"] = gdf["primary_educ"].astype(float)
gdf = gdf[gdf["cohort"].notna() & gdf["primary_educ"].notna()].copy()

# Fixed scale across all maps
vmin = gdf["primary_educ"].min()
vmax = gdf["primary_educ"].max()

for cohort in sorted(gdf["cohort"].unique()):
    subset = gdf[gdf["cohort"] == cohort].copy()
    values = subset["primary_educ"].dropna().values

    fig, ax = plt.subplots(figsize=(8, 10))

    # ---- map ----
    subset.plot(
        column="primary_educ",
        cmap="viridis",
        linewidth=0.4,
        edgecolor="black",
        legend=False,
        vmin=vmin,
        vmax=vmax,
        ax=ax
    )

    ax.set_title(f"Mali birthplace-based primary education, cohort {cohort}")
    ax.set_axis_off()

    # ---- horizontal colorbar ----
    norm = mpl.colors.Normalize(vmin=vmin, vmax=vmax)
    sm = mpl.cm.ScalarMappable(norm=norm, cmap="viridis")
    sm.set_array([])

    cbar = fig.colorbar(
        sm,
        ax=ax,
        orientation="horizontal",
        fraction=0.05,
        pad=0.04
    )
    cbar.set_label("Primary education share")

    # ---- wave axis above the colorbar ----
    # Get the colorbar position in figure coordinates
    cb_pos = cbar.ax.get_position()

    # Add a small axis directly above the colorbar
    wave_ax = fig.add_axes([
        cb_pos.x0,                # left
        cb_pos.y1 + 0.005,        # bottom
        cb_pos.width,             # width
        0.05                      # height
    ])

    mean_val = values.mean()
    min_val = values.min()
    max_val = values.max()

    fig.text(
        0.5, 0.02,
        f"Mean: {mean_val:.3f}   Min: {min_val:.3f}   Max: {max_val:.3f}",
        ha='center',
        fontsize=10
    )

    # Histogram-based smooth-ish wave
    kde = gaussian_kde(values)
    x = np.linspace(vmin, vmax, 300)
    y = kde(x)

    wave_ax.plot(x, y, linewidth=1.5)
    wave_ax.fill_between(x, y, alpha=0.25)

    # Match wave axis to colorbar scale
    wave_ax.set_xlim(vmin, vmax)
    wave_ax.set_xticks([])
    wave_ax.set_yticks([])

    # Clean appearance
    for spine in wave_ax.spines.values():
        spine.set_visible(False)

    out_file = output_dir / f"birthplace_primary_educ_{cohort}.png"
    plt.savefig(out_file, dpi=300, bbox_inches="tight")
    plt.close()

print("Done. Maps saved to:", output_dir)
```
